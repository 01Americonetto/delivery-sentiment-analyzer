```python
# 1. Instala a biblioteca do Hugging Face (Transformers)
!pip install -q transformers torch

from transformers import pipeline

# 2. Inicializa o pipeline de análise de sentimento
print("Carregando o modelo de IA...")
analisador = pipeline("sentiment-analysis")

# 3. Criamos uma lista com avaliações simuladas de clientes (estilo iFood)
avaliacoes_clientes = [
    "A comida chegou super rápido, quentinha e o entregador foi muito educado!",
    "O pedido demorou mais de uma hora e veio faltando o refrigerante.",
    "Experiência razoável, a comida é boa mas o tempo de entrega deixou a desejar."
]

# 4. Rodando a IA para classificar o feedback
print("\n--- RELATÓRIO DE ANÁLISE DE SENTIMENTO DOS PEDIDOS ---\n")
for i, review in enumerate(avaliacoes_clientes, 1):
    resultado = analisador(review)[0]
    print(f"Pedido {i}: '{review}'")
    print(f"👉 Classificação: {resultado['label']} (Confiança: {resultado['score']:.2f})\n")
```
